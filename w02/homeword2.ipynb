Bài 10

```python
import math
x0 = float(input("Nhập x0 (tâm x): "))
y0 = float(input("Nhập y0 (tâm y): "))
r  = float(input("Nhập bán kính r: "))
x = float(input("Nhập x (điểm A): "))
y = float(input("Nhập y (điểm A): "))
d2 = (x - x0)**2 + (y - y0)**2
r2 = r**2
if math.isclose(d2, r2, rel_tol=1e-9):
    print("Điểm A nằm TRÊN đường tròn.")
elif d2 < r2:
    print("Điểm A nằm TRONG đường tròn.")
else:
    print("Điểm A nằm NGOÀI đường tròn.")
```
