bai 13

```python
A = float(input("Nhập A: "))
tong = 0
n = 0

while tong <= A:
    n += 1
    tong += 1 / n

print("n nhỏ nhất là:", n)
```

bai 14

```python
x = int(input("Nhập số: "))
max_val = x
min_val = x

while x != -1:
    if x > max_val:
        max_val = x
    if x < min_val:
        min_val = x
    x = int(input("Nhập số: "))

print("Số lớn nhất:", max_val)
print("Số nhỏ nhất:", min_val)
```

bai 16

```python
n = int(input("Nhập n: "))
chan = 0
le = 0

while n > 0:
    cs = n % 10
    if cs % 2 == 0:
        chan += 1
    else:
        le += 1
    n //= 10

print("Chữ số chẵn:", chan)
print("Chữ số lẻ:", le)
```

bai 18

```python
n = int(input("Nhập n: "))

while n % 3 == 0:
    n //= 3

if n == 1:
    print("Là số dạng 3^k")
else:
    print("Không phải dạng 3^k")
```

bai 19

```python
A = int(input("Nhập A: "))

f1 = 1
f2 = 1

while f2 <= A:
    f1, f2 = f2, f1 + f2

print("Số Fibonacci lớn nhất không vượt A là:", f1)
```

```python
n = int(input("Nhập n: "))
dem = 0

while n > 0:
    cs = n % 10
    if cs == 2 or cs == 3 or cs == 5 or cs == 7:
        dem += 1
    n //= 10

print("Số chữ số nguyên tố:", dem)
```
