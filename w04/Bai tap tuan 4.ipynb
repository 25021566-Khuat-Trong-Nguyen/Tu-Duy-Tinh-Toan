```python
while True :
    n = float(input("nhap so nguyen: "))
    while n != float(n):
        n = int(input("nhap lai so nguyen: "))
    if n < 0:
        print("no la so am")
    if n < 2 and n >= 0:
        print("ko phai so nguyen to")
    elif n >= 2 and n == int(n):
        nguyen_to = True
        for i in range(2, int(n**0.5) + 1 ):
            if n % i == 0:
                nguyen_to = False
        if nguyen_to:
            print(n,"la so nguyen to")
            break
        else:
            print(n,"ko phai so nguyen to")
            break
             

```
