```python
import math
def checkPrime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

n = int(input("Nhập số nguyên dương n: "))
if checkPrime(n):
    print(f"{n} là số nguyên tố")
else:
    print(f"{n} không phải là số nguyên tố")
```

```python
def sumOfDigits(n: int) -> int:
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
n = int(input("Nhập số nguyên dương n: "))
print(f"Tổng các chữ số của {n} là: {sumOfDigits(n)}")
```

```python
def word_frequency(sentence):
    words = sentence.split()
    freq = {}

    for word in words:
        if word in freq:
            freq[word] += 1
        else:
            freq[word] = 1

    return freq
sentence = input("Nhập câu: ")
print(word_frequency(sentence))
```

```python
students = {'S001' : {  'name': 'An', 'score': 20},
           'S002' : {  'name': 'Binh', 'score': 21}
}
students['S003'] = {'name': 'Cuong', 'score': 22}
for sid, info in students.items():
    print(sid, info['name'], "-", info['score'])

total_score = 0
for info in students.values():
    total_score += info['score']

average = total_score / len(students)
print("Điểm trung bình là:", average)
```

```python
def binary_research(arr, value):
    left,right = 0,len(arr) -1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == value:
            return mid
        elif arr[mid] < value:
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = [1,3,5,7,9,11]
value = 9
print(binary_research(arr, value))
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        sx = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                sx = True
        if not sx:
            break
    return arr
arr = list(map(int,input().split()))
print(bubble_sort(arr))
```
